````r
SCRIPT 3: EXPORTAR ANÁLISES PARA PDF
# ============================================================

library(ggplot2)
library(gridExtra)
library(pdftools)  # Para manipulação de PDF

# Criar função para gerar gráficos
gerar_graficos_zika <- function(dados) {

  # Gráfico 1: Série temporal
  p1 <- dados %>%
    mutate(mes_notif = floor_date(as.Date(DT_NOTIFIC), "month")) %>%
    count(mes_notif) %>%
    ggplot(aes(x = mes_notif, y = n)) +
    geom_line(color = "#2c7fb8", size = 1.2) +
    geom_point(color = "#2c7fb8", size = 2) +
    labs(title = "Evolução Temporal dos Casos de Zika",
         x = "Mês da Notificação", y = "Número de Casos") +
    theme_minimal() +
    theme(plot.title = element_text(hjust = 0.5, face = "bold"))

  # Gráfico 2: Casos por UF
  p2 <- dados %>%
    count(SG_UF_NOT, name = "casos") %>%
    top_n(10, casos) %>%
    ggplot(aes(x = reorder(SG_UF_NOT, casos), y = casos, fill = casos)) +
    geom_col() +
    coord_flip() +
    scale_fill_gradient(low = "#fee090", high = "#d73027") +
    labs(title = "Top 10 UFs com Maior Número de Notificações",
         x = "UF", y = "Número de Casos") +
    theme_minimal() +
    theme(legend.position = "none")

  # Gráfico 3: Distribuição por sexo
  p3 <- dados %>%
    filter(CS_SEXO %in% c("M", "F")) %>%
    count(CS_SEXO) %>%
    mutate(percentual = n/sum(n)*100) %>%
    ggplot(aes(x = CS_SEXO, y = n, fill = CS_SEXO)) +
    geom_col() +
    geom_text(aes(label = paste0(round(percentual, 1), "%")),
              vjust = -0.5, size = 5) +
    scale_x_discrete(labels = c("F" = "Feminino", "M" = "Masculino")) +
    labs(title = "Distribuição por Sexo", x = "", y = "Número de Casos") +
    theme_minimal() +
    theme(legend.position = "none")

  # Gráfico 4: Casos em gestantes
  p4 <- dados %>%
    filter(CS_GESTANT %in% c("1", "2", "3", "4", "5")) %>%
    count(CS_GESTANT) %>%
    mutate(
      gestacao = case_when(
        CS_GESTANT == "1" ~ "1º Trimestre",
        CS_GESTANT == "2" ~ "2º Trimestre",
        CS_GESTANT == "3" ~ "3º Trimestre",
        CS_GESTANT == "4" ~ "Idade gestacional ignorada",
        CS_GESTANT == "5" ~ "Não gestante"
      )
    ) %>%
    ggplot(aes(x = reorder(gestacao, -n), y = n, fill = gestacao)) +
    geom_col() +
    coord_flip() +
    labs(title = "Casos de Zika em Mulheres em Idade Fértil",
         subtitle = "Por condição gestacional",
         x = "", y = "Número de Casos") +
    theme_minimal() +
    theme(legend.position = "none")

  # Gráfico 5: Classificação final
  p5 <- dados %>%
    count(CLASSI_FIN) %>%
    mutate(
      classificacao = case_when(
        CLASSI_FIN == "1" ~ "Confirmado",
        CLASSI_FIN == "2" ~ "Descartado",
        TRUE ~ "Outros"
      ),
      percentual = n/sum(n)*100
    ) %>%
    ggplot(aes(x = "", y = percentual, fill = classificacao)) +
    geom_bar(stat = "identity", width = 1) +
    coord_polar("y", start = 0) +
    geom_text(aes(label = paste0(round(percentual, 1), "%")),
              position = position_stack(vjust = 0.5)) +
    labs(title = "Proporção de Casos Confirmados") +
    theme_void()

  return(list(p1, p2, p3, p4, p5))
}

# Gerar e salvar em PDF
salvar_pdf_analises <- function(dados, arquivo = "analises_zika.pdf") {
  graficos <- gerar_graficos_zika(dados)

  pdf(arquivo, width = 12, height = 8)

  # Página 1: Título
  plot.new()
  text(0.5, 0.7, "RELATÓRIO DE ANÁLISE", cex = 2.5, font = 2)
  text(0.5, 0.5, "Dados de Zika Vírus - SINAN", cex = 1.5)
  text(0.5, 0.3, paste("Data de geração:", Sys.Date()), cex = 1)

  # Página 2: Série temporal e mapa
  grid.arrange(graficos[[1]], graficos[[2]], ncol = 2)

  # Página 3: Perfil demográfico
  grid.arrange(graficos[[3]], graficos[[4]], ncol = 2)

  # Página 4: Classificação
  grid.arrange(graficos[[5]])

  # Página 5: Tabela resumo
  plot.new()
  grid.table(
    dados %>%
      summarise(
        `Total de Notificações` = n(),
        `Casos Confirmados` = sum(CLASSI_FIN == "1", na.rm = TRUE),
        `Taxa de Confirmação` = paste0(round(sum(CLASSI_FIN == "1", na.rm = TRUE)/n()*100, 1), "%"),
        `Óbitos` = sum(EVOLUCAO == "2", na.rm = TRUE),
        `Letalidade` = paste0(round(sum(EVOLUCAO == "2", na.rm = TRUE)/sum(CLASSI_FIN == "1", na.rm = TRUE)*100, 2), "%")
      )
  )

  dev.off()
  cat("PDF salvo em:", arquivo, "\n")
}

# Executar
salvar_pdf_analises(dados_selecionados)
````
